## What is a Stack?
A Stack is a linear data structure that follows the LIFO (Last In, First Out) principle. The element that is added last is the one removed first.



### Q1: What is a Stack?
- A stack is a linear data structure based on LIFO. The last element inserted is the first to be removed.
- Real-life Examples of Stack:
    - A pile of plates: The last plate placed is the first to be removed.

    - Browser history: You go back to the most recent page first.

- Stack Operations (CRUD):
    - Operation	Description
    - push	Add an element to the top
    - pop	Remove the top element
    - peek or top	View the top element without removing it
    - is_empty	Check if the stack is empty

### Q2: How is Stack different from Queue?
- Stack → LIFO
- Queue → FIFO
- Stack removes from top; Queue removes from front.

### Q3: Can we implement a Stack using Queues?
- Yes. Using two queues:
- Push: Enqueue to the empty queue, then dequeue all from the other.
- Pop: Just dequeue from the front.

### Q4: Where are stacks used in programming?
- Function call/recursion stack

- Expression evaluation
 
- Undo mechanisms
 
- Backtracking algorithms
 
- Depth First Search (DFS)

### Q5: What’s the space complexity of a stack?
- O(n), where n is the number of elements in the stack.

### Q6: Can we use arrays to implement stack?
- Yes, arrays/lists can be used with append() and pop().

### Q7: What happens if we pop from an empty stack?
- It usually raises an Underflow Error or IndexError.

### Q8: What is Stack Overflow?
- Stack Overflow happens when the stack exceeds its maximum limit, often in recursion without base case.

### Q9: Is stack memory static or dynamic?
- It depends. In low-level languages, it’s often static. In Python (high-level), it's dynamic (like lists grow dynamically).

## Implementation
### 1. Using `collections.deque`

```python
from collections import deque

stack = deque()

stack.append(10)
stack.append(20)
stack.append(30)
stack.append(40)
stack.append(50)

def pop():
    return stack.pop()

def peek():
    return stack[-1]

def is_empty():
    return len(stack) == 0


print("Stack: ",stack)
print(pop())
print("Updated Stack", stack)
print("Is Empty", is_empty())
print("Peek", peek())
```

### 2. Stack using custom class

```python
class Stack:
    def __init__(self):
        self.stack = []

    def peek(self)-> int:
        self.stack[-1]
    
    def is_empty(self)-> bool:
        return len(self.stack) == 0
    
    def add(self, x: int) -> None:
        self.stack.append(x)
    
    def pop(self)-> int:
        if self.is_empty():
            raise IndexError("Pop from empty stack")
        return self.stack.pop()
    
    def size(self) -> int:
        return len(self.stack)
    
    def display(self) -> None:
        for i in self.stack:
            print(i, end=" ")
        print()

# Initialize Stack
st1 = Stack()
st1.add(10)
st1.add(20)
st1.add(30)
st1.add(40)
st1.add(50)
st1.add(60)
# Print Stack
st1.display()

# Is stack Empty?
print("Is stack empty:", st1.is_empty())

# Peek Element
print("Peek:", st1.peek())

# Pop Element
print("Pop:", st1.pop())

# Size of Stack
print("Size", st1.size())
```

### 3. Stack Using Linked List

```python
class Node:
    def __init__(self, data: int):
        self.data = data
        self.next = None

class StackUsingLL:
    def __init__(self):
        self.top = None
    
    def is_empty(self):
        return self.top == None

    def add(self, data: int):
        new_node = Node(data)
        if self.top == None:
            self.top = new_node
        else:
            temp =  self.top
            self.top = new_node
            new_node.next = temp
    
    def pop(self) -> int:
        if not self.is_empty():
            temp = self.top
            self.top = temp.next
            temp.next = None
            return temp.data
        else:
            print("Stack is empty")
        
    def peek(self) -> int:
        return self.top.data
    
    def display(self):
        if self.is_empty():
            print("Stack is empty")
            return
        temp = self.top
        while temp != None:
            print(temp.data, end=" ")
            temp = temp.next
        print()

st1 = StackUsingLL()
st1.add(10)
st1.add(20)
st1.add(30)
st1.add(40)
st1.add(50)
st1.add(60)
print("Display:", end=" ")
st1.display()
print("Top element:",st1.peek())
print("Before pop:", end=" ")
st1.display()
st1.pop()
print("After pop:", end=" ")
st1.display()
```
