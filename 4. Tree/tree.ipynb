# 1. 🌳**General Tree**

### 🔹 Q1: What is a Tree?

A **tree** is a non-linear hierarchical data structure made up of **nodes**, with one node designated as the **root**, and others forming **branches or subtrees**.


### 🔹 Q2: What is a General Tree?

A **General Tree** is a tree in which **each node can have any number of children** (not limited to 2 like a binary tree).

Example:

```
        A
      / | \
     B  C  D
        |
        E
```


### 🔹 Q3: What are the key terms?

| Term    | Meaning                                       |
| ------- | --------------------------------------------- |
| Node    | An individual element in the tree             |
| Root    | The topmost node (no parent)                  |
| Parent  | Node with children                            |
| Child   | Node connected below a parent                 |
| Leaf    | Node with no children                         |
| Sibling | Nodes with the same parent                    |
| Subtree | A tree formed by any node and its descendants |
| Height  | Longest path from root to a leaf              |
| Depth   | Distance from root to a node                  |
| Degree  | Number of children a node has                 |



### 🔹 Q4: Is a Binary Tree a type of General Tree?

✅ Yes. A **binary tree** is a special case of a general tree where **each node has at most two children** (left and right).


### 🔹 Q5: Where is a General Tree used?

* XML/HTML document structure
* File system hierarchy
* AI (decision trees)
* DOM in web development
* Org charts, family trees



### i. Using Classes

````python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

    def add_child(self, child_node):
        self.children.append(child_node)

    def display(self, level=0):
        print(" " * level + str(self.data))
        for child in self.children:
            child.display(level + 1)
    
root = TreeNode("A")
child1 = TreeNode("B")
child2 = TreeNode("C")
child3 = TreeNode("D")
child2_1 = TreeNode("E")

root.add_child(child1)
root.add_child(child2)
root.add_child(child3)
child2.add_child(child2_1)

root.display()
````

### ii. Using Python List

````python
tree = ["A", [["B"], ["C", [["E"]]], ["D"]]]
def display_list_tree(node, level=0):
    if isinstance(node, list):
        print(" " * level + str(node[0]))
        if len(node) > 1:
            for child in node[1]:
                display_list_tree(child, level + 1)

display_list_tree(tree)
````

# 🌳 2. Binary Tree
### Q1: What is a Binary Tree?
A Binary Tree is a hierarchical data structure where each node has at most two children, referred to as left and right.

### Q2: What are the different types of Binary Trees?

| Type                     | Description                                                 |
| ------------------------ | ----------------------------------------------------------- |
| **Full Binary Tree**     | Every node has 0 or 2 children                              |
| **Complete Binary Tree** | All levels are full except possibly the last (left-aligned) |
| **Perfect Binary Tree**  | All levels completely filled, and all leaves at same depth  |
| **Balanced Binary Tree** | Height difference between left and right subtrees ≤ 1       |
| **Degenerate Tree**      | Each node has only one child (like a linked list)           |


🔹 Q3: What are common Binary Tree operations?
- Traversal (visit nodes in an order):

    - Inorder (Left → Root → Right)

    - Preorder (Root → Left → Right)

    - Postorder (Left → Right → Root)

    - Level Order (BFS)

- Insert (for complete trees)

- Search (recursive)

- Delete (only in complete binary tree or heap-like structures)

- Height/Depth

- Leaf node count

- Check symmetry/mirror

### Q4: Applications of Binary Trees?
- Abstract syntax trees (compilers)

- Expression evaluation trees

- Binary Search Trees (BST)

- File system hierarchies

- Huffman encoding

- Routing tables (decision trees)

### Q5: Time & Space Complexities (general)
| Operation | Time Complexity | Space Complexity |
| --------- | --------------- | ---------------- |
| Insert    | O(n)            | O(n)             |
| Search    | O(n)            | O(n)             |
| Traversal | O(n)            | O(n)             |


````python
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinaryTree:
    def __init__(self):
        self.root = None
    
    def build_from_level_order(self, arr):
        if not arr or arr[0] is None:
            self.root = None
            return
        
        self.root = BinaryTreeNode(arr[0])
        queue = [self.root]
        i = 1

        while queue and i < len(arr):
            current = queue.pop(0)
            if i < len(arr) and arr[i] is not None:
                current.left = BinaryTreeNode(arr[i])
                queue.append(current.left)
            i += 1

            if i < len(arr) and arr[i] is not None:
                current.right = BinaryTreeNode(arr[i])
                queue.append(current.right)
            i += 1

    def inorder(self, node):
        if node:
            self.inorder(node.left)
            print(node.data, end=" ")
            self.inorder(node.right)
    
    def preorder(self, node):
        if node:
            print(node.data, end=" ")
            self.preorder(node.left)
            self.preorder(node.right)

    def postorder(self, node):
        if node:
            self.postorder(node.left)
            self.postorder(node.right)
            print(node.data, end=" ")
    
    def level_order(self):
        if not self.root:
            return
        queue = [self.root]
        while queue:
            node = queue.pop(0)
            print(node.data, end=" ")
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

arr = ["A", "B", "C", "D", "E", "F", "G"]
btree = BinaryTree()
btree.build_from_level_order(arr)

print("Inorder Traversal:")
btree.inorder(btree.root)

print("\nPreorder Traversal:")
btree.preorder(btree.root)

print("\nPostorder Traversal:")
btree.postorder(btree.root)
````
