

# 🔗📃 LINKED LIST


### 🔹 Q1: What is a Linked List?

A **Linked List** is a linear data structure where elements are stored in **nodes**, and each node points to the next node. Unlike arrays, elements are not stored in contiguous memory.

### 🔹 Q2: Why use Linked List over an Array?

| Feature                | Array             | Linked List                |
| ---------------------- | ----------------- | -------------------------- |
| Size                   | Fixed or resizing | Dynamic                    |
| Access Time            | O(1) (via index)  | O(n) (via traversal)       |
| Insert/Delete (middle) | O(n)              | O(1) (if pointer is known) |

### 🔹 Q3: What are the types of Linked Lists?

* **Singly Linked List** – points to the next node.
* **Doubly Linked List** – points to both next and previous.
* **Circular Linked List** – last node points to the head.
* **Doubly Circular Linked List** – both directions + circular.



### 🔹 Q4: What is the structure of a node?

Each node typically has:

* `data`: the value
* `next`: pointer to the next node
* `prev`: (in doubly linked list) pointer to the previous node


### 🔹 Q5: What are the key operations?

* `insert_at_head(value)`
* `insert_at_tail(value)`
* `insert_at_position(index, value)`
* `delete(value)` or `delete_at_position(index)`
* `search(value)`
* `display()` or `traverse()`


### 🔹 Q6: What are real-life uses of Linked Lists?

* Implementation of stacks and queues
* Memory management (e.g., free list in OS)
* Navigation systems (previous/next steps)
* Blockchain-like structures


### 🔹 Q7: What are the time complexities?

| Operation      | Singly Linked List |
| -------------- | ------------------ |
| Insert at head | O(1)               |
| Insert at tail | O(n)               |
| Search         | O(n)               |
| Delete node    | O(n)               |
| Traverse       | O(n)               |


## Types of Linked List
### 1. Singly Linked List

```python
class Node():
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self) -> bool:
        return self.head == None
    
    def insert_at_head(self, data: int) -> None:
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def insert_at_tail(self, data: int) -> None:
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
        
    def search(self, target: int) -> bool:        
        temp = self.head
        while temp:
            if temp.data == target:
                return True
            temp = temp.next
        return False

    def delete(self, target: int) -> None:
        temp = self.head
        if not temp:
            return
        
        if temp.data == target:
            self.head = temp.next
            return
        prev = None
        while temp and temp.data != target:
            prev = temp
            temp = temp.next
        if temp:
            prev.next = temp.next

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" ->")
            temp = temp.next
        print("None")

ll = LinkedList()
ll.insert_at_head(10)
ll.insert_at_tail(20)
ll.insert_at_tail(30)
ll.insert_at_tail(40)
ll.insert_at_tail(50)
ll.display()

print("Search:", ll.search(50))
ll.delete(30)
ll.display()
print("Linked List is print:", ll.is_empty())
```

### 2. Doubly Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0
    
    def is_empty(self):
        return self.size == 0
    
    def insert_at_end(self, data: int)-> None:
        new_node = Node(data=data)
        if self.is_empty():
            self.head = new_node
        else:
            curr = self.head
            while curr.next:
                curr = curr.next
            curr.next = new_node
            new_node.prev = curr
        self.size += 1

    def insert_at_head(self, data: int):
        new_node = Node(data=data)
        new_node.next = self.head
        self.head = new_node
        self.head.prev = new_node
        self.size+=1

    def insert_after(self, data: int, key: int):
        new_node = Node(data=data)
        curr = self.head
        if curr == key:
            curr.prev = new_node
            new_node.next = curr
            self.head = new_node
            self.size += 1
            return

        
        while curr.next is not None:
            if curr.data == key:
                # Getting adjacent nodes
                next_node = curr.next
                prev_node = curr
                # Linking previous node
                prev_node.next = new_node
                new_node.prev = prev_node
                # Linking next node
                next_node.prev = new_node
                new_node.next = next_node
                self.size += 1
                return
            curr = curr.next

        if curr.data == key:
            curr.next = new_node
            new_node.prev = curr
            self.size += 1
        else: 
            print("Node not Found")

    
    def update(self, key, new_data):
        curr = self.head
        while curr:
            if curr.data == key:
                curr.data = new_data
                return
            curr = curr.next
        
    def get_size(self):
        return self.size

    def peek(self):
        return self.head.data

    def display(self):
        curr = self.head
        while curr:
            print(curr.data, end="<->")
            curr = curr.next
        print("None")


dll = DoublyLinkedList()
dll.insert_at_head(10)
dll.insert_at_end(20)
dll.insert_at_end(30)
dll.insert_at_end(40)
dll.insert_after(data=25, key=20)
dll.display()
dll.update(key=25, new_data=21)
print("After Update:", end="")
dll.display()
print("Size of Doubly Linked List: ",dll.get_size())
```

### 3. Circular Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def is_empty(self) -> bool:
        return self.size == 0
    
    def append(self, data) -> None:
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            new_node.next = new_node
            self.size += 1
            return
        curr = self.head.next
        while curr.next != self.head:
            curr = curr.next
        new_node.next = curr.next
        curr.next = new_node
        self.size += 1
        return
    
    def peek(self) -> int:
        if self.is_empty():
            return
        return self.head.data

    def get_size(self) -> int:
        return self.size

    def get_head(self) -> int:
        if self.is_empty():
            return
        return self.head.data

    def insert_at_head(self, data) -> None:
        new_node = Node(data=data)
        if self.is_empty():
            self.head = new_node
            new_node.next = new_node
            self.size += 1
            return
        curr = self.head.next
        while curr.next != self.head:
            curr = curr.next
        curr.next = new_node
        new_node.next = self.head.next
        self.head = new_node
        self.size += 1
        return
    
    def insert_at_position(self, data, position) -> None:
        if position < 0:
            return
        elif position == 0:
            self.insert_at_head(data=data)
            return
        new_node = Node(data=data)
        curr = self.head
        prev = None
        count = 0
        while count < position - 1:
            prev = curr
            curr = curr.next
            if curr == self.head:
                break
            count += 1
        prev.next = new_node
        new_node.next = curr
        self.size += 1

    def delete(self, key) -> None:
        if self.is_empty():
            print("Circular Linked List is already empty!")
            return
        curr = self.head
        if self.head.data == key:
            while curr.next != self.head:
                curr = curr.next
            self.head = self.head.next
            curr.next = self.next
            self.size -= 1
            return
        prev = None
        while curr.next != self.head:
            prev = curr
            curr = curr.next
            if curr.data == key:
                prev.next = curr.next
                curr.next = None
                self.size -=1
                return
        print("Data Not found")
        return

    def update(self, key, new_data) -> None:
        if self.is_empty():
            print("Circular Linked List is already empty!")
            return
        curr = self.head
        if curr.data == key:
            curr.data = new_data
            return
        while curr.next != self.head:
            curr = curr.next
            if curr.data == key:
                curr.data = new_data
                return
        print("Data Not Found")
        return

    def display(self) -> None:
        if self.is_empty():
            print("Circular Linked List is already empty!")
            return

        print(self.head.data, end="->")
        curr = self.head.next
        while curr != self.head:
            print(curr.data, end="->")
            curr = curr.next
        print(None)
cll = CircularLinkedList()
cll.append(10)
cll.append(20)
cll.append(30)
cll.append(40)
cll.append(50)
cll.append(60)
cll.insert_at_position(data = 66, position=7)
cll.display()
cll.delete(20)
cll.display()
cll.update(key=40, new_data=45)
print("After Update")
cll.insert_at_head(100)
cll.display()
```
