## 1. Bubble Sort
### ✅ Logic
- Repeatedly compare adjacent elements.

- Swap them if they’re in the wrong order.

- Continue this process for all elements.

- Largest element "bubbles up" to the end in each pass.

Do this n-1 times (for n elements).

### 🧠 Time Complexity
- Worst/Avg: O(n²)

- Best (already sorted): O(n) — with optimization

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
            if not swapped:
                break
    return arr

print(bubble_sort(arr))
```

## 2. Selection Sort
### ✅ Logic
- Divide array into sorted and unsorted parts.

- Repeatedly find the minimum from unsorted part.

- Swap it with the first unsorted element.

- Expand the sorted part one element at a time.

### 🧠 Time Complexity
Always: O(n²)

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_index = i
        for j in range(i+1, n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
```

```python
selection_sort(arr)
```

## 3. Insertion Sort
### ✅ Logic (Note-Taking Style)
- Start from the second element.

- Pick an element and place it in its correct position among the previous elements.

- Like sorting cards in hand.

### 🧠 Time Complexity
- Worst/Avg: O(n²)

- Best (nearly sorted): O(n)

```python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i -1
    
        while j >=0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr
```

```python
insertion_sort(arr)
```

## 4. Merge Sort
### ✅ Logic
- Divide and Conquer algorithm.

- Recursively divide the array into halves until you get single-element arrays.

- Then merge those arrays in a sorted way.

- Repeat until the whole array is merged and sorted.

### 🧠 Time & Space Complexity
| Case    | Time       | Space |
| ------- | ---------- | ----- |
| Best    | O(n log n) | O(n)  |
| Average | O(n log n) | O(n)  |
| Worst   | O(n log n) | O(n)  |


- Stable Sort
- Not in-place (uses extra space)

```python
arr = [4,2,1,6,8,3]
def merge_sort(arr):
    if len(arr) <=1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result

merge_sort(arr)
```

## 5. Quick Sort
### ✅ Logic
- Pick a pivot.
- Partition array so:
    - All elements < pivot go left
    - All elements > pivot go right

Recursively apply to left and right subarrays.
### 🧠 Time & Space Complexity
| Case  | Time       | Space    |
| ----- | ---------- | -------- |
| Best  | O(n log n) | O(log n) |
| Worst | O(n²)      | O(log n) |
| Avg   | O(n log n) | O(log n) |
- In-place (if implemented properly)
- Not stable by default

```python
arr = [4,2,1,6,8,3]
def quick_sort(arr):
    if len(arr) <=1:
        return arr
    
    pivot = arr[-1]
    left = [x for x in arr[:-1] if x <= pivot]
    right = [x for x in arr[:-1] if x > pivot]

    return quick_sort(left) + [pivot] + quick_sort(right)
quick_sort(arr)
```

```python
arr = [4,2,1,6,8,3]
def quicksort(arr, left, right):
    if left < right:
        partition_pos = partition(arr, left, right)
        quicksort(arr, left, partition_pos - 1)
        quicksort(arr, partition_pos + 1, right)

def partition(arr, left, right):
    i = left
    j = right -1
    pivot = arr[right]

    while i < j:
        while i < right and arr[i] < pivot:
            i += 1
        while j > left and arr[j] >= pivot:
            j -= 1
        
        if i < j:
            arr[i], arr[j] = arr[j], arr[i]
    
    if arr[i] > pivot:
        arr[i], arr[right] = arr[right], arr[i]

    return i

quicksort(arr, 0, len(arr)-1)
print(arr)
```

## 6. Heap Sort
### ✅ Logic
- Use a max heap (binary tree where parent ≥ children).

- Build a heap from the array.

- Repeatedly extract the max and rebuild the heap.

### 🧠 Time & Space Complexity
Case	Time	Space
All	O(n log n)	O(1)

- In-place
- Not stable

```python
arr = [4,2,1,6,8,3]
def heapify(arr, n, i):
    largest = i
    left = 2*i + 1
    right = 2*i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right
    
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    for i in range(n//2 - 1, -1, -1):
        heapify(arr, n, i)
    
    for i in range(n-1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)
    return arr

heap_sort(arr)
```

## 7. Counting Sort (Non-comparison)
### ✅ Logic
- Count occurrences of each number.

- Use counts to determine sorted position.

### 🧠 Time & Space Complexity
O(n + k), where k = range of values

- Only works with non-negative integers in known range

```python
arr = [2, 1, 1, 3, 0]
def counting_sort(arr):
    max_val = max(arr)
    count = [0] * (max_val + 1)

    for num in arr:
        count[num] += 1
    
    result = []
    for i,c in enumerate(count):
        result.extend([i]* c)
    return result

print(counting_sort(arr))
```

## 8. Radix Sort (Non-comparison)
### ✅ Logic
- Sort numbers digit by digit (from least significant to most).

- Use Counting Sort as a subroutine.

### 🧠 Time Complexity
- O(nk), where k is number of digits

```python
arr = [772, 16, 81, 28, 11, 156]
def counting_sort_by_digit(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 10
    
    for num in arr:
        index = (num // exp) % 10
        count[index] += 1

    for i in range(1, 10):
        count[i] += count[i -1]
    
    for i in reversed(range(n)):
        index = (arr[i] // exp) % 10
        output[count[index] - 1] = arr[i]
        count[index] -= 1
    
    for i in range(n):
        arr[i] = output[i]

def radix_sort(arr):
    max_num = max(arr)
    exp = 1
    while max_num // exp > 0:
        counting_sort_by_digit(arr, exp)
        exp *= 10
    return arr

print(radix_sort(arr))
```

## 9. Bucket Sort (for floats or uniform distribution)
## ✅ Logic
- Divide range into buckets

- Sort each bucket (often with Insertion Sort)

- Concatenate buckets

```python
arr = [40,10,22,89,1,55]
def bucket_sort(arr):
    if len(arr) == 0:
        return []
    
    min_val = min(arr)
    max_val = max(arr)
    n = len(arr)

    if max_val == min_val:
        return arr.copy()

    buckets = [[] for _ in range(n)]
    
    for num in arr:
        index = int((num - min_val) / (max_val - min_val) * (n-1))
        buckets[index].append(num)
    
    for bucket in buckets:
        bucket.sort()

    result = []

    for bucket in buckets:
        result.extend(bucket)
    
    return result

print(bucket_sort(arr))
    
```
