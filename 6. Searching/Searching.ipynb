## 1. 🔎 Linear Search
### ✅ Logic:
- Traverse the array from start to end.

- Compare each element with the target.

- If found, return its index.

- If the loop ends, return -1 (not found).

## ⏱ Time & Space Complexity:
- Time: O(n)



```python
arr= [1, 6, 8, 9, 5]
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
print(linear_search(arr, 9))
```

```python
arr= [1, 5, 6, 8, 9, 15, 55, 77]
def binary_search(arr, target):
    low, high = 0, len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid -1
    return -1

print(binary_search(arr, 77))

```

## ⚡ 2. Binary Search
### ✅ Logic:
- Only works on sorted arrays.

- Find the middle element.

- If target < mid → search left.

- If target > mid → search right.

- Repeat until found or low > high.

### ⏱ Time & Space Complexity:
- Time: O(log n)

- Space:

    - Iterative: O(1)

    - Recursive: O(log n) (due to call stack

```python
arr= [1, 5, 6, 8, 9, 15, 55, 77]
def binary_search_recursive(arr, target, low, high):
    if low > high:
        return -1
    mid = (low + high) // 2

    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid+1, high)
    else:
        return binary_search_recursive(arr, target, low, mid-1)

print(binary_search_recursive(arr, 77, 0, len(arr)-1))
```

## 🦘 3. Jump Search
### ✅ Logic:
- Works on sorted arrays.

- Jump ahead by √n steps until the block may contain the target.

- Do a linear search within that block.

### ⏱ Time & Space Complexity:
- Time: O(√n)

- Space: O(1)

```python
import math
arr= [1, 5, 6, 8, 9, 15, 55, 77]
def jump_search(arr, target):
    n = len(arr)
    step = int(math.sqrt(n))
    prev = 0

    while prev < n and arr[min(step, n) - 1] < target:
        prev = step
        step += int(math.sqrt(n))
        if prev >= n:
            return -1
    
    for i in range(prev, min(step, n)):
        if arr[i] == target:
            return i
    return -1

print(jump_search(arr, 15))
```

## 📈 4. Interpolation Search
### ✅ Logic:
- Works best for uniformly distributed sorted arrays.

- Estimate position using interpolation formula.

- Adjust range based on comparison.

### ⏱ Time & Space Complexity:
- Time:
    - Best: O(log log n)

    - Worst: O(n)

- Space: O(1)

```python
arr= [1, 5, 6, 8, 9, 15, 55, 77] # More than 1 step
# arr= [1, 3, 5, 7, 9, 11, 13, 15] # 1 step
def interpolation_search(arr, target):
    low, high = 0, len(arr) - 1

    while low <= high and arr[low] <= target <= arr[high]:
        if arr[high] == arr[low]:
            if arr[low] == target:
                return low
            return -1

        pos = low + ((target - arr[low]) * (high - low) // (arr[high] - arr[low]))
        
        if pos < 0 or pos >= len(arr):
            return -1
        if arr[pos] == target:
            return pos
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1
    return -1

interpolation_search(arr, 15)
```

## 🚀 5. Exponential Search
### ✅ Logic:
- Works on sorted arrays.

- Start at index 1, and double the index until element is greater than or equal to the target.

- Perform binary search between the found bounds.

### ⏱ Time & Space Complexity:
- Time: O(log i) → i is the position where target lies

- Space: O(1) (if using iterative binary search)

```python
def binary_search(arr, target, left, right):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
        
    return -1

def exponential_search(arr, target):
    if arr[0] == target:
        return 0
    
    i = 1
    while i < len(arr) and arr[i] <= target:
        i *= 2
    return binary_search(arr, target, i //2, min(i, len(arr) - 1))
```
