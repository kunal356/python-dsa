# QUEUE


### Q1: What is a Queue?

A **Queue** is a linear data structure that follows the **FIFO (First In, First Out)** principle. The first element inserted is the first to be removed.


### Q2: Why use a Queue?

* Models real-world lines (e.g., printer queue, task scheduling)
* Used in **BFS**, caching, resource management, etc.


### Q3: What are the types of Queues?

| Type               | Description                                     |
| ------------------ | ----------------------------------------------- |
| **Simple Queue**   | Basic FIFO                                      |
| **Circular Queue** | End connects to front to use space efficiently  |
| **Deque**          | Double-ended queue (insert/delete at both ends) |
| **Priority Queue** | Elements are dequeued based on priority         |

### Q4: What are the basic operations?

| Operation    | Description                        |
| ------------ | ---------------------------------- |
| `enqueue(x)` | Add element to rear                |
| `dequeue()`  | Remove element from front          |
| `front()`    | Peek the front element             |
| `is_empty()` | Check if queue is empty            |
| `size()`     | Return number of elements in queue |


### Q5: Where is a Queue used?

* Print/job scheduling
* OS-level task queues
* Level-order traversal in trees (BFS)
* Message passing in systems

### Q6: Time Complexity

| Operation  | Time Complexity |
| ---------- | --------------- |
| Enqueue    | O(1)            |
| Dequeue    | O(1)            |
| Front/Peek | O(1)            |
| is\_empty  | O(1)            |

## 1. Simple Queue:
 #### 1. Using Custom Class

````python
class Queue():
    def __init__(self):
        self.head = None
        self.size = 0
    
    def is_empty(self) -> bool:
        return self.size == 0
    
    def get_size(self) -> int:
        return self.size
    
    def peek(self) -> int:
        return self.head.data
    
    def enqueue(self, data) -> None:
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            self.size += 1
            return
        curr = self.head
        while curr.next != None:
            curr = curr.next
        curr.next = new_node
        self.size += 1

    def dequeue(self) -> None:
        self.head = self.head.next
        self.size -= 1
        

    def display(self) -> None:
        curr = self.head
        while curr != None:
            print(curr.data, end="->")
            curr = curr.next
        print("None")
q1 = Queue()
q1.enqueue(10)
q1.enqueue(20)
q1.enqueue(30)
q1.enqueue(40)
q1.display()
print("Size before dequeue:",q1.get_size())
q1.dequeue()
print("Queue after dequeue:",end=" ")
q1.display()
print("Size after dequeue:",q1.get_size())
````

## 2. 🔁Circular Queue

### Q1: What is a Circular Queue?

A **Circular Queue** is a linear data structure in which the last position connects back to the first, forming a **circle**. It efficiently uses storage by reusing freed space.



### Q2: Why Circular Queue over a regular queue?

In a simple queue (array-based), after some dequeues, front moves forward but space at the beginning is wasted.
Circular Queue solves this by wrapping around.



### Q3: How is it implemented?

Using:

* A fixed-size array
* Two pointers: `front` and `rear`
* Wrap around with modulo (`%`) operator



### Q4: When is the queue full?

**Full Condition:**

```python
(front == (rear + 1) % size)
```

### Q5: When is the queue empty?

**Empty Condition:**

```python
(front == -1)
```
### Q6: Time Complexity

| Operation | Time Complexity |
| --------- | --------------- |
| Enqueue   | O(1)            |
| Dequeue   | O(1)            |
| Peek      | O(1)            |

````python
class CircularQueue():
    def __init__(self, size):
        self.size = size
        self.queue = [None] * size
        self.front = -1
        self.rear = -1
    
    def enqueue(self, value) -> None:
        if self.front == (self.rear + 1) % self.size:
            print("Queue is full")
            return
        
        if self.front == -1:
            self.front = 0
        self.rear = (self.rear + 1) % self.size
        self.queue[self.rear] = value

    def dequeue(self) -> int:
        if self.front == -1:
            print("Queue is empty")
            return
        
        removed = self.queue[self.front]
        if self.front == self.rear:
            self.front = self.rear = -1
        else:
            self.front = (self.front + 1) % self.size
        return removed
    
    def peek(self) -> int:
        if self.front == -1:
            return None
        return self.queue[self.front]
    
    def display(self):
        if self.front == -1:
            print("Queue is empty")
            return
        i = self.front
        while True:
            print(self.queue[i], end=" ")
            if i == self.rear:
                break
            i = (i + 1) % self.size
        print()
        


c1 = CircularQueue(4)
c1.enqueue(1)
c1.enqueue(2)
c1.enqueue(3)
c1.enqueue(4)
c1.display()
print("Front element", c1.peek())
print("Deleted element:", c1.dequeue())
print("Queue after deque:",end=" ")
c1.display()
c1.enqueue(5)
c1.display()
````

## 3. DEQUE (Double-Ended Queue)

### Q1: What is a Deque?

A **Deque** (pronounced *"deck"*) is a linear data structure that allows insertion and deletion from **both the front and the rear**.



### Q2: What are the main operations?

| Operation            | Description       |
| -------------------- | ----------------- |
| `append(x)`          | Insert at rear    |
| `appendleft(x)`      | Insert at front   |
| `pop()`              | Remove from rear  |
| `popleft()`          | Remove from front |
| `peek()` / `[-1]`    | View rear         |
| `peekleft()` / `[0]` | View front        |



### Q3: What are the types of Deque?

| Type                        | Description                            |
| --------------------------- | -------------------------------------- |
| **Input-restricted deque**  | Insertion at one end, deletion at both |
| **Output-restricted deque** | Deletion at one end, insertion at both |


### Q4: Where is Deque used?

* Sliding window problems
* Palindrome checking
* Undo/redo operations
* BFS (level-order traversal)
* Browser history


### Q5: Time Complexity

| Operation         | Time Complexity |
| ----------------- | --------------- |
| Insert Front/Back | O(1)            |
| Delete Front/Back | O(1)            |
| Peek              | O(1)            |



````python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

class Deque:
    def __init__(self):
        self.front = None
        self.rear = None

    def peek(self)->int:
        return self.front.data
    
    def append(self, data):
        new_node = Node(data)
        if not self.rear:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            new_node.prev = self.rear
            self.rear = new_node
    
    def append_left(self, data):
        new_node = Node(data)
        if not self.rear:
            self.front = self.rear = new_node
        else:
            self.front.prev = new_node
            new_node.next = self.front
            self.front = new_node

    def pop(self):
        if not self.rear:
            print("Deque is already empty")
            return
        data = self.rear.data
        self.rear = self.rear.prev
        if self.rear:
            self.rear.next = None
        else:
            self.front = None
        return data
    
    def pop_left(self):
        if not self.front:
            print("Deque is already empty")
            return
        data = self.front.data
        self.front = self.front.next
        if self.front:
            self.front.prev = None
        else:
            self.rear = None
        return data
    
    def display(self):
        temp = self.front
        while temp:
            print(temp.data, end=" <-> ")
            temp = temp.next
        print("None")


d1 = Deque()
d1.append(1)
d1.append(2)
d1.append(3)
d1.append(4)
d1.append_left(5)
d1.display()
print("Pop",d1.pop())
d1.display()
print("Pop Left",d1.pop_left())
d1.display()
print("Front element:",d1.peek())
````


## 4. PRIORITY QUEUE

### Q1: What is a Priority Queue?

A **Priority Queue** is an abstract data type similar to a regular queue, but where **each element has a priority**. Elements are dequeued based on **priority, not just order of arrival**.

### Q2: How does it differ from a regular queue?

| Feature   | Queue | Priority Queue                |
| --------- | ----- | ----------------------------- |
| Order     | FIFO  | Based on priority             |
| Insertion | End   | Anywhere                      |
| Deletion  | Front | Element with highest priority |


### Q3: What are the main operations?

| Operation      | Description                                     |
| -------------- | ----------------------------------------------- |
| `insert(x, p)` | Insert element `x` with priority `p`            |
| `pop()`        | Remove and return element with highest priority |
| `peek()`       | View highest-priority element                   |
| `is_empty()`   | Check if queue is empty                         |



### Q4: How is Priority Queue implemented?

1. **Using Sorted List** – Insert sorted → O(n), pop is O(1)
2. **Using Heap (Best)** – Both insert and delete are O(log n)
3. **Using `queue.PriorityQueue`** (built-in Python)
4. **Using `heapq` module** – Lightweight and fast



### Q5: Min Heap vs Max Heap

* **Min Heap**: smallest value has highest priority (default in Python)
* **Max Heap**: largest value has highest priority (requires negation trick)



### Q6: Where are Priority Queues used?

* CPU scheduling (shortest job first)
* Dijkstra’s algorithm (shortest path)
* Huffman coding
* Event simulation
* A\* search algorithm



### Q7: Time Complexity with Heap

| Operation | Time Complexity |
| --------- | --------------- |
| Insert    | O(log n)        |
| Pop       | O(log n)        |
| Peek      | O(1)            |

````python
class PriorityQueue:
    def __init__(self):
        self.heap = []
    
    def insert(self, priority, data):
        self.heap.append((priority, data))
        self._heapify_up(len(self.heap) - 1)
    
    def pop(self):
        if self.is_empty():
            print("Priority Queue is empty!")
            return
        self._swap(0, len(self.heap) - 1)
        item = self.heap.pop()
        self._heapify_down(0)
        return item[1]
    
    def peek(self):
        if self.is_empty():
            return None
        return self.heap[0][1]
    
    def is_empty(self):
        return len(self.heap) == 0
    
    def _parent(self, i):
        return (i - 1) // 2
    
    def _left(self, i):
        return 2 * i + 1
    
    def _right(self, i):
        return 2 * i + 2
    
    def _swap(self, i, j):
        self.heap[i], self.heap[j] = self.heap[j], self.heap[i]

    def _heapify_up(self, index):
        parent = self._parent(index)
        if index > 0 and self.heap[index][0] < self.heap[parent][0]:
            self._swap(index, parent)
            self._heapify_up(parent)

    def _heapify_down(self, index):
        smallest = index
        left = self._left(index)
        right = self._right(index)

        if left < len(self.heap) and self.heap[left][0] < self.heap[smallest][0]:
            smallest = left

        if right < len(self.heap) and self.heap[right][0] < self.heap[smallest][0]:
            smallest = right

        if smallest != index:
            self._swap(index, smallest)
            self._heapify_down(smallest)

    def display(self):
        print([item[1] for item in self.heap])

pq = PriorityQueue()
pq.insert(3, "C")
pq.insert(1, "A")
pq.insert(2, "B")
pq.insert(0, "Urgent")

print("Top priority:", pq.peek())
print("All items by priority:")
pq.display()
````

### 2. Using `heapq` module

````python
import heapq

class PriorityQueue:
    def __init__(self):
        self.heap = []

    def insert(self, data, priority):
        heapq.heappush(self.heap, (priority,data))

    def pop(self):
        if not self.is_empty():
            return heapq.heappop(self.heap)[1]
    
    def is_empty(self):
        return len(self.heap) == 0
    
    def peek(self):
        if not self.is_empty():
            return self.heap[0][1]
    
    def display(self):
        print([item[1] for item in self.heap])

pq1 = PriorityQueue()
pq1.insert(data=1, priority=0)
pq1.insert(data=2, priority=1)
pq1.insert(data=3, priority=2)
print("Is Empty",pq1.is_empty())
print("Peek",pq1.peek())
pq1.display()
print(pq1.pop())
````
